```python
for fold, train, valid in cv_folds(df):
    model = DecisionTreeClassifier(max_depth=3, random_state=42)
    model.fit(train[X], train[TARGET])

    predictions = model.predict(valid[X])
    print(fold, evaluate(valid[TARGET], predictions))
```
